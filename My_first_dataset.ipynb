```python
import numpy as np
data=np.genfromtxt("/content/drive/MyDrive/Exam_Score_Prediction.csv",dtype=None, delimiter=',', skip_header=1, names=('student_id','age','gender','course','study_hour','class_attendance','internet_access','sleep_hour','sleep_quality','study_method','facility_rating','exam_difficulty','exam_score'))
```

```python
data['exam_score']=(data['exam_score']/50)*100
data['exam_score']
```

```python
import re
grades='ACAAAABCBCBAA'
print(re.findall("B",grades))
print(re.findall("[AB]",grades))
print(re.findall("[A][B-C]",grades))

print(re.findall("AB|AC",grades))
print(re.findall("[^A]",grades))
print(re.findall("^[^A]", grades))
print(re.findall("^[^B]", grades))
```

```python
print(re.findall("A{2,10}",grades))
print(re.findall("A{3,3}A{1,1}", grades))
print(re.findall("A{2,3}A{1,1}", grades))
print(re.findall("A{2,2}A{1,1}", grades))
print(re.findall("A{1,2}A{1,1}", grades))#B2B max 3A's to min 2A's
print(re.findall("A{1,1}A{1,1}", grades))#back-to-back 2A's
```
